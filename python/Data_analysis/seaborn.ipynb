# Line Plot

```python
import seaborn as sns
import matplotlib.pyplot as plt

var = [1,2,3,4,5,6,7]
var_1 = [2,3,4,1,5,6,7]

plt.plot(var, var_1)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

var = [1,2,3,4,5,6,7]
var_1 = [2,3,4,1,5,6,7]

x_1 = pd.DataFrame({"var": var, "var_1": var_1})

sns.lineplot(x="var", y = "var_1", data = x_1)

plt.show()
```

```python
var = [1,2,3,4,5,6,7]
var_1 = [2,3,4,1,5,6,7]

sns.lineplot(x=var, y=var_1)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

y_1 = sns.load_dataset("penguins").head(20)

sns.lineplot(x="bill_length_mm", y ="flipper_length_mm", data=y_1,hue="sex", style="sex", palette="Accent", markers=["o",">"], dashes=False,
             legend="full" )
plt.grid()
plt.title("python seaborn ")
plt.show()
```

# Histogram Plot

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

var = sns.load_dataset("penguins")
sns.displot(var["bill_length_mm"], kde=True, rug=True, color="green",
            log_scale=True)
plt.show()
```

# scatter plot

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd


var = sns.load_dataset("penguins").head(20)
sns.scatterplot(x="bill_length_mm", y = "bill_depth_mm", data=var, hue="sex", style="sex")
plt.show()
```

# Heatmap

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np
data = sns.load_dataset("anagrams")
var = np.linspace(1,10,20).reshape(4,5)
sns.heatmap(var)

```

```python
import numpy as np
import seaborn as sns
ar = np.array([["a0", "a1", "a2", "a3", "a4"],
                ["b0", "b1", "b2", "b3", "b4"]])

# data = sns.load_dataset("anagrams")
var = np.linspace(1,10,20).reshape(4,5)

sns.heatmap(var, vmin=0, vmax=10, cmap="PuOr", annot=ar, fmt="s")
```
