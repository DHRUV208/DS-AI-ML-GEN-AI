# BAR PLOT

```python
import matplotlib.pyplot as plt
x = ["python", "c", "cpp", "Java"]
y = [85, 70, 60, 82]
plt.xlabel("lang", fontsize=15)
plt.ylabel("Number",  fontsize=15)
plt.title("python libr")
c = ["y", "b", "m", "g"]
z = [20,30,40,50]
plt.bar(x,y,
        width=0.5,
        color="r",
        align="center",
        edgecolor="black",
        linewidth=2,
        linestyle=":",
        alpha=0.8,
        label="brainworks" 
        )
plt.bar(
        x,z,
        width=0.5,
        color="y",
        align="center",
        edgecolor="black",
        linewidth=2,
        linestyle=":",
        alpha=0.8,
        label="brainworks1"
)
plt.legend()
plt.show()
```

# Scatter Plot

```python
import matplotlib.pyplot as plt
day = [1,2,3,4,5,6,7]
num = [2,3,1,4,5,3,6]
# colors=["r", "y", "g", "b", "r", "g", "b"]
colors = [10,49,30,29,56,20,30]
sizes = [500,200,400,500,800,1000,700]
plt.scatter(day,
            num,
            c=colors,
            s=sizes,
            marker="*",
            edgecolor="g",
            linewidth=1,
            cmap="viridis"
            )
plt.colorbar()
plt.title("Scatter Plot", fontsize=20)
plt.xlabel("Day")
plt.ylabel("No.")
plt.show()
```

# step plot

```python
import matplotlib.pyplot as plt
x = [1,2,3,4,5]
y = [2,2,3,5,4]
plt.step(x,y)
plt.title("python")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid()
plt.show()
```

# fill between

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1,2,3,4,5])
y = np.array([1,2,3,4,5])

plt.plot(x,y)
plt.fill_between(x,y,color="g", where=(x>=2) & (x<=4))
plt.title("python")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```
