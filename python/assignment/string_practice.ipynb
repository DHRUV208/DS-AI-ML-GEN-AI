```python

def checkAnagram(str1, str2):
    if(len(str1) != len(str2)):
        return "not anagram"
    for x in str1:
        if (x.isalpha()):
            if (str1.count(x) != str2.count(x)):
                return "not anagram"

    return "anagram"
    


s1_input = input("enter string1")
s2_input = input("enter string2")
s1 = s1_input.lower()
s2 = s2_input.lower()
res = checkAnagram(s1, s2)
print(res)
```

```python
# resetting password

def resetPassword(passw1, passw2):
    if(passw1 == passw2):
        return "password updted successfully"
    elif(passw1.casefold() == passw2.casefold()):
        return "check cases & try again"
    else:
        return "password do not match"



passw_input = input("enter passw")
passw_reinput = input("enter passw again")
res = resetPassword(passw_input, passw_reinput)
print(res)
```

```python
# Wage Calculator
def calculateWage(hours, wages):
    total_hrs = sum(hours)
    # total_wages
    if(total_hrs<=40):
        total_wages = total_hrs * wages
    else:
        overtime = total_hrs - 40
        total_wages = 40* wages + overtime* wages * 1.5

    return total_wages




hours = input("enter hours: ")
wage = int (input("enter hourly wage: "))

hours = hours.split()

week_hours = [int(x) for x in hours]
res = calculateWage(week_hours, wage)
print(res)





```

```python
# inverted dict

def invertedDict(original_dict):
    inverted = {}
    for key, value in original_dict.items():
        if value not in inverted:
            inverted[value] = key
        else:
            inverted[value].add(key)
        
    return inverted



original_dictionary = {'a': (1,2), 'b': 2, 'c': 3, 'd': 7}
res = invertedDict(original_dictionary)
print(res)
```

```python
def findDuplicates(nums):
    seen, dups = set(), []
    for n in nums:
        if n in seen and n not in dups:
            dups.append(n)
        seen.add(n)
    return dups

findDuplicates([1,2,3,4,2,4,4,1])
```

```python
#  sort without built in 
import random as rd


list_comp = [rd.randint(10, 50) for i in range(20)]
print("before sorting: ", list_comp)
n = len(list_comp)
for i in range(n):
    for j in range(n-1-i):
        if list_comp[j] > list_comp[j+1]:
            list_comp[j], list_comp[j+1] = list_comp[j+1], list_comp[j]
        
    
print("after sorting: ", list_comp)
```
