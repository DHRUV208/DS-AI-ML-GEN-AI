```python

def checkAnagram(str1, str2):
    if(len(str1) != len(str2)):
        return "not anagram"
    for x in str1:
        if (x.isalpha()):
            if (str1.count(x) != str2.count(x)):
                return "not anagram"

    return "anagram"
    


s1_input = input("enter string1")
s2_input = input("enter string2")
s1 = s1_input.lower()
s2 = s2_input.lower()
res = checkAnagram(s1, s2)
print(res)
```

```python
# Data cleaning project

def data_cleaning(str_input):
    clean = ""
    for x in str_input:
        if(x.isalpha()):
            clean += x
        else:
            clean += " "

    return clean


scan = "These+notes^reveal#some*cool stuffs))"
res = data_cleaning(scan)
print(res)

```

```python
# resetting password

def resetPassword(passw1, passw2):
    if(passw1 == passw2):
        return "password updted successfully"
    elif(passw1.casefold() == passw2.casefold()):
        return "check cases & try again"
    else:
        return "password do not match"



passw_input = input("enter passw")
passw_reinput = input("enter passw again")
res = resetPassword(passw_input, passw_reinput)
print(res)
```

```python
# Wage Calculator
def calculateWage(hours, wages):
    total_hrs = sum(hours)
    # total_wages
    if(total_hrs<=40):
        total_wages = total_hrs * wages
    else:
        overtime = total_hrs - 40
        total_wages = 40* wages + overtime* wages * 1.5

    return total_wages




hours = input("enter hours: ")
wage = int (input("enter hourly wage: "))

hours = hours.split()

week_hours = [int(x) for x in hours]
res = calculateWage(week_hours, wage)
print(res)





```

```python
# REMOVE DUPLICATES


def removeDuplicates(l):
    res = []
    for element in l:
        if element not in res:
            res.append(element)

    return res


l1_input = input("enter numbers: ")

# l1_input = l1_input.split()
l1_input = [int(x) for x in l1_input.split()]


# l1 = list(l1_input)
print("l1_input: ", l1_input)

res = removeDuplicates(l1_input)
print(res)
```

```python
# rotate a list

def rotateList(list, n):

    rotated = list[n:] + list[:n]
    return rotated


list_input = input("enter numbers")
n = int (input("number of rotations"))


list_input = [int(x) for x in list_input.split()]

res = rotateList(list_input, n)
print(res)
```

```python
# pallindrome list
def pallindromeList(listt):
    rev_list = listt[::-1]
    if (listt == rev_list):
        return "pallindrome yess"
    else:
        return "pallindrome noo"
    

list_input = input("enter list numbers")

list_input = [int(x) for x in list_input.split()]

res = pallindromeList(list_input)
print(res)
```

```python
# itertools permutations & combinations
import itertools as it

def permutations(listt):
    perms = it.permutations(listt, r = 2)
    print("perms: ", perms)
    perm_list = list(perms)

    print("Permutations: ", perm_list)
    for t in perm_list:
        print(t)
    return perm_list


list_input = input("enter list char")
list_input = [x for x in list_input]
print("input: ", list_input)
res = permutations(list_input)
print("res: ", res)
```

```python
# Split odd even

def splitOddEven(listt):
    odd_list = [x for x in listt if x%2 != 0 ]
    even_list = [x for x in listt if x%2 == 0 ]

    return f'odd list = {odd_list}\n even list = {even_list}'


list_input = input("Enter list numbers: ")
list_input = [int(x) for x in list_input.split()]

res = splitOddEven(list_input)
print(res)
```

```python
# find longest list
def longestList(listt):
    max_list = max(listt, key = len)
    return max_list



list_input = input("enter list numbers")
list_input = [int(x) for x in list_input.split()]

res = longestList(list_input)
print(res)
```

```python
# `Problem-10`:Write a python program that receives a list of strings and performs bag of word operation on those strings
def buck(x):
    s = 1
    for i in x:
        l = i.split()
        print("l: ", l)
        d = {}
        for i in l:
            if i in d:
                d[i] += 1
            else:
                d[i] = 1
        print(f'bag{s} =',d)
        s += 1

x = ['cat mat rat cat cat sat rat vat',
     'sat fat cat fat sat vat bat cat fat ',
     'pat cat mat rat']
buck(x)

```
