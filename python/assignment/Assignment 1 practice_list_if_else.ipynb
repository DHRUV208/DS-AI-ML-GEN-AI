```python
# Check whether a number is even or odd
def checkNumIsEvenOrOdd(num):
    if(num == 0):
        return "neither odd nor even"
    elif(num%2 == 0):
        return "even"
    else:
        return "odd"

var = int(input("enter a number: "))
res = checkNumIsEvenOrOdd(var)
print(res)
```

```python
# Find the largest of two numbers
def findLargest(a,b):
    if(a == b):
        return f"{a} is equal to {b}"
    elif(a>b):
        return f"{a} is largest"
    else:
        return f"{b} is largest"

p = int(input("enter a number: "))
q = int(input("enter a number: "))
res = findLargest(p,q)
print(res)
```

```python
# Find the largest of three numbers
def findLargest(a,b,c):
    if(a == b == c):
        return f"{a} is equal {b} is equal to {c}"
    elif(a > b > c):
        return f"{a} is largest"
    elif( a < b > c):
        return f"{b} is largest"
    elif ((a < b < c) | (a > b < c)):
        return f"{c} is largest"
# 8 5 9   4 2 6
p= int(input("enter a number: "))
q = int(input("enter a number: "))
r = int(input("enter a number: "))
res = findLargest(p,q,r)
print(res)
    
```

```python
# Check whether a year is a leap year

def leapYear(year):
    if(year % 100 == 0):
        if(year % 400 == 0):
            return "leap year"
        else:
            return "not a leap year"
    elif(year % 4 == 0):
        return "leap year"
    else:
        return "not a leap year"

var = int(input("enter year: "))
res = leapYear(var)
print(res)
```

```python
# Check whether a triangle is equilateral, isosceles, or scalene
def checkTriangle(a,b,c):
    if(a == b == c):
        return "Equilateral Triangle"
    elif((a == b) | (b == c) | (a == c)):
        return "Isosceles Triangle"
    elif( a != b & b != c & a !=c):
        return "Scalene Triangle"

p = int(input("enter number: "))
q = int(input("enter number: "))
r = int(input("enter number: "))
res = checkTriangle(p,q,r)
print(res)
```

```python
# Simple calculator (+, -, *, /) using if-elif-else
def calculator (a,b, operator):
    if(operator == '+'):
        return a + b
    elif(operator == '-'):
        return a-b
    elif(operator == '*'):
        return a*b
    elif(operator == '/'):
        if(b == 0):
            return "division-by-zero-error"
        else:
            return a/b
    else:
        return "invalid input"

p = int(input("enter number"))
q = int(input("enter number"))
operator_input=input("enter operator: ")
res = calculator(p,q,operator_input)
print(res)
```

```python
# Calculate an electricity bill with slab rates

# 0–100 units: ₹2 per unit
# 101–200 units: ₹3 per unit
# 201–300 units: ₹5 per unit
# Above 300 units: ₹7 per unit

def electricityBill(units):
    bill=0
    if(units<=100):
        bill = units*2
    elif(units<=200):
        bill = 100*2 + (units-100)*3
    elif(units<=300):
        bill = 100*2 + 100 *3 + (units-200)* 5
    else:
        bill = 100*2 + 100*3 + 100*5 + (units-300)*7

    return bill


units_input = int(input("enter unit: "))
res = electricityBill(units_input)
print(res)
    

```

```python


# Check whether a character is uppercase, lowercase, digit, or special
def checkChar(char):
    if(char.isupper()):
        return "Uppercase"
    elif(char.islower()):
        return "Lowercase"
    elif(char.isdigit()):
        return "Digit"
    else:
        return "Special Character"


char_input = input("enter char: ")
res = checkChar(char_input)
print(res)
```

```python
# Calculate income tax by salary slab

# Salary up to ₹3,00,000: 0% tax
# ₹3,00,001–₹6,00,000: 5% tax
# ₹6,00,001–₹9,00,000: 10% tax
# Above ₹9,00,000: 15% tax


def incomeTaxCalculate(salary):
    tax=0
    if(salary < 0):
        return "invalid input"
    elif(salary<=300000):
        tax=0
    elif(salary<=600000):
        tax = (salary-300000)* 0.05
    elif(salary<=900000):
        tax = (300000*0.05) + (salary - 600000) * 0.10
    else:
        tax = (300000 * 0.05) + (300000 * 0.10) + (salary - 900000)*0.15

    if salary>=0:
        print(f"Annual salary: ${salary:,.2f}")
        print(f"Income tax: ${tax:,.2f}")
        print(f"Salary after tax: ${salary-tax:,.2f}")

annual_salary = int(input("enter annual salary: "))

res = incomeTaxCalculate(annual_salary)   
```
