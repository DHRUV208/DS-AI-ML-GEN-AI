```python
class Employee:
    def __init__(self, first, last, salary):
        self.first = first 
        self.last = last 
        self.email = "{}.{}@company.com".format(self.first,self.last)
        self.salary = salary 

    def return_name(self):
        return "{} {}".format(self.first, self.last)

    def apply_raise(self):
        self.salary = self.salary * 1.04


emp_1 = Employee("Ram", "Patil", 50000)
emp_2 = Employee("Sham", "Patil", 60000)

print(emp_1.salary)
emp_1.apply_raise()
print(emp_1.salary)
```

```python
class Employee:
    raise_amount = 1.04

    def __init__(self, first, last, salary):
        self.first = first 
        self.last = last 
        self.email = "{}.{}@company.com".format(self.first,self.last)
        self.salary = salary 

    def return_name(self):
        return "{} {}".format(self.first, self.last)

    def apply_raise(self):
        self.salary = self.salary * self.raise_amount


emp_1 = Employee("Ram", "Patil", 50000)
emp_2 = Employee("Sham", "Patil", 60000)

print(emp_1.salary)
emp_1.apply_raise()
print(emp_1.salary)
print(Employee.raise_amount)

print("-"*50)

print(emp_1.__dict__)
print(Employee.__dict__)

print("-"*50)

Employee.raise_amount = 1.05

print(Employee.raise_amount)
print(emp_1.salary)
emp_1.apply_raise()
print(emp_1.salary)
```

```python
class Employee:
    raise_amount = 1.04
    no_emp = 0

    def __init__(self, first, last, salary):
        self.first = first 
        self.last = last 
        self.email = "{}.{}@company.com".format(self.first,self.last)
        self.salary = salary 
        Employee.no_emp += 1

    def return_name(self):
        return "{} {}".format(self.first, self.last)

    def apply_raise(self):
        self.salary = self.salary * self.raise_amount


emp_1 = Employee("Ram", "Patil", 50000)
emp_2 = Employee("Sham", "Patil", 60000)

print(Employee.no_emp)
```
